# Example 6.1: Ewald Sphere Construction

## Overview
This example demonstrates the **Ewald sphere construction**, a fundamental tool in electron diffraction for determining which reciprocal lattice points (reflections) are in the Bragg condition.

## Key Concepts Illustrated:
1. **Relativistic electron wavelength** calculation for different energies
2. **Ewald sphere radius** ($k = 1/\lambda$) dependence on energy
3. **Reciprocal lattice** construction for a cubic crystal
4. **Interactive visualization** showing which reflections can be simultaneously excited
5. **Excitation error** calculation for off-Bragg conditions

## Physical Insight:
The Ewald sphere radius increases with electron energy (shorter wavelength), allowing access to higher-angle reflections. This is the basis for choosing electron energy in TEM experiments.

## Connection to Lectures:
- Used in Lecture 6.1 for zone-axis diffraction setup
- Foundation for understanding Bloch-wave excitation conditions
- Determines which beams couple in the structure matrix (Lecture 6.2)

```python
import numpy as np

try:
    import plotly.graph_objects as go
except ImportError as exc:
    raise ImportError("Install plotly to enable interactive 2D: pip install plotly") from exc

# Electron wavelength with relativistic correction.
def electron_wavelength_angstrom(energy_keV):
    h = 6.62607015e-34  # J*s
    m_e = 9.10938356e-31  # kg
    c = 299792458.0  # m/s
    e = 1.602176634e-19  # C
    energy_J = energy_keV * 1.0e3 * e
    denom = np.sqrt(2.0 * m_e * energy_J * (1.0 + energy_J / (2.0 * m_e * c**2)))
    wavelength_m = h / denom
    return wavelength_m * 1.0e10  # m -> Å

a_angstrom = 3.0
hkl_max = 4

# Reciprocal lattice points (simple cubic).
hkl = np.arange(-hkl_max, hkl_max + 1)
H, K, L = np.meshgrid(hkl, hkl, hkl, indexing="ij")
Gx = (H / a_angstrom).ravel()
Gy = (K / a_angstrom).ravel()
Gz = (L / a_angstrom).ravel()

# Use the ky=0 plane for a 2D cross-section.
plane_mask = np.isclose(Gy, 0.0)
Gx2 = Gx[plane_mask]
Gz2 = Gz[plane_mask]

def ewald_circle(k, n=300):
    t = np.linspace(0, 2 * np.pi, n)
    x = k * np.cos(t)
    z = k * np.sin(t) + k
    return x, z

energies_keV = [50, 75, 100, 125, 150, 175, 200]
max_g = hkl_max / a_angstrom
x_lim = max_g * 1.2
z_lim = max_g * 1.2

frames = []
for energy in energies_keV:
    lam = electron_wavelength_angstrom(energy)
    k = 1.0 / lam
    cx, cz = ewald_circle(k)
    frames.append(
        go.Frame(
            name=str(energy),
            data=[
                go.Scatter(x=cx, y=cz, mode="lines"),
            ],
        )
    )

initial_energy = energies_keV[2]
lam0 = electron_wavelength_angstrom(initial_energy)
k0 = 1.0 / lam0
cx0, cz0 = ewald_circle(k0)

fig = go.Figure(
    data=[
        go.Scatter(x=cx0, y=cz0, mode="lines", line=dict(color="#4c72b0", width=2), name="Ewald circle"),
        go.Scatter(x=Gx2, y=Gz2, mode="markers", marker=dict(size=7, color="#1f1f1f"), name="Reciprocal lattice"),
        go.Scatter(x=[0], y=[0], mode="markers", marker=dict(size=10, color="#d62728"), name="Origin"),
    ],
    frames=frames,
    layout=go.Layout(
        title=f"Ewald circle (ky=0): {initial_energy} keV",
        template="plotly_white",
        xaxis=dict(title="g_x (1/Å)", range=[-x_lim, x_lim], zeroline=True),
        yaxis=dict(title="g_z (1/Å)", range=[0, z_lim], scaleanchor="x", scaleratio=1),
        margin=dict(l=60, r=20, t=40, b=50),
        updatemenus=[
            dict(
                type="buttons",
                showactive=False,
                x=0.02,
                y=0.98,
                buttons=[
                    dict(label="Play", method="animate", args=[None, {"frame": {"duration": 600, "redraw": True}, "fromcurrent": True}]),
                    dict(label="Pause", method="animate", args=[[None], {"frame": {"duration": 0, "redraw": False}, "mode": "immediate"}]),
                ],
            )
        ],
        sliders=[
            dict(
                active=2,
                currentvalue={"prefix": "Energy (keV): "},
                pad={"t": 20},
                steps=[
                    {
                        "label": str(energy),
                        "method": "animate",
                        "args": [[str(energy)], {"frame": {"duration": 0, "redraw": True}, "mode": "immediate"}],
                    }
                    for energy in energies_keV
                ],
            )
        ],
    ),
)

fig.show()

print("Problem 1: Ewald Sphere Construction")
print("=" * 50)
```
