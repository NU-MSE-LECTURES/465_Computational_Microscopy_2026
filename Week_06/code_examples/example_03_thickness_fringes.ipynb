## Physical Insight:
Pendellösung arises from beat interference between two Bloch waves. The oscillation period depends on the extinction distance, which is inversely proportional to the structure factor. Thicker crystals show more oscillations.

## Key Equations to Remember

1. **Bragg Law**: $2d\sin\theta_B = n\lambda$ (n=1 for electrons)
2. **Extinction Distance**: $\xi_g = \lambda / (2|\chi_g|)$ 
3. **Excitation Error**: $s_g = \mathbf{g}\cdot\hat{\mathbf{k}}_i + |\mathbf{g}|^2/(2k)$
4. **Effective Extinction Distance**: $\xi_{eff} = \xi_g / \sqrt{1 + (s_g\xi_g)^2}$
5. **Thickness Fringe**: $I_g(t) = \sin^2(\pi t/\xi_{eff})\exp(-t/\Lambda_{abs})$
