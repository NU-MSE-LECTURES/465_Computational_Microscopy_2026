## Key Concepts

1. **Bloch Theorem**: $\psi_{nk}(x) = e^{ikx} u_{nk}(x)$ where $u$ has crystal periodicity
2. **Plane-Wave Expansion**: $\psi = \sum_g C_g e^{i(k+g)x}$
3. **Hamiltonian in Reciprocal Space**: H = kinetic energy + potential matrix
4. **Band Structure**: Eigenvalues $E_n(k)$ show energy dispersion
5. **Avoided Crossing**: Bands repel each other due to potential coupling
6. **Channeling**: Low-energy states localized to potential minima
7. **Absorption**: Complex potential → complex eigenvalues → finite lifetime
