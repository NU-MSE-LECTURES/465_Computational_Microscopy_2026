## Key Equations Summary

**Kinematic**:
$$I_g = |F_g|^2 \text{ (constant with thickness)}$$

**Dynamical** (Two-Beam):
$$I_g(t) = \sin^2(\pi t / \xi_{eff}) \exp(-t/\Lambda_{abs})$$

**Extinction Distance**:
$$\xi_g = \frac{\lambda}{2|\chi_g|}$$

**Excitation Error**:
$$s_g = \mathbf{g} \cdot \hat{\mathbf{k}}_i + |\mathbf{g}|^2/(2k)$$
